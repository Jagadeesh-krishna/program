```python
import numpy as np
```

```python
# addition
mat1=np.array([[1,2],[3,4]])
mat2=np.array([[5,6],[7,8]])
result=mat1+mat2
mul=mat1*mat2
print(mul)
print("addition:",result)
```

```python
result=np.matmul(mat1, mat2)
print(result)
result = mat1 @mat2
print(result)
result=np.dot(mat1,mat2)
print(result)
```

```python
result = mat1/mat2
print(result)
```

```python
# transpose of matrix
mat=np.array([[1,2,3],[4,5,6]])
print(mat.T)
```

```python
print(mat.shape)
```

```python
identity=np.eye(5)
print(identity)
```

```python
# determinant
mat=np.array([[1,2],[3,4]])
print(np.linalg.det(mat))
```

```python
# matrix inverse
mat=np.array([[1,2],[3,4]])
print(np.linalg.inv(mat))
```
