```python
import pandas as pd
import numpy as np
```

```python
# series
arr=np.array([10,20,30,40,50])
marks=pd.Series(arr)
print(marks)
```

```python
marks=pd.Series(
    [100,100,100],
    index=["maths", "science", "english"]
)
print(marks)
print(marks["maths"])
print(marks.shape)
print(marks.size)
print(marks.dtype)
print(marks.values)
print(marks.index)
```
