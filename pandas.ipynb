```python
import pandas as pd
import numpy as np
```

```python
# series
arr=np.array([10,20,30,40,50])
marks=pd.Series(arr)
print(marks)
```

```python
marks=pd.Series(
    [100,100,100],
    index=["maths", "science", "english"]
)
print(marks)
print(marks["maths"])
print(marks.shape)
print(marks.size)
print(marks.dtype)
print(marks.values)
print(marks.index)
```

```python
numbers=[1,2,34,7]
result=[]
for i in numbers:
  result.append(i*2)
print(result)
```

```python
# vectorized operation means applying one operation to all elements automatically without writing loops
numbers=[1,2,34,7]
arr=np.array(numbers)
print(arr*2)
```

```python
data=pd.Series([1,2,3,4,np.nan,5])
print(data.isnull()) #to know the missing value
print(data.fillna(0)) #repleace the null value with any value
```

```python
# dataframe
data={
    "name":["A","B", "C","D" ],
    "marks":[60,70,80,90],
    "grade":["b","b","a", "a"]
}
df=pd.DataFrame(data)
print(df)
print(df.shape)
print(df.size)
print(df.columns)
print(df.index)
print(df.tail)
```

```python
df["City"]=["tirupati", "renigunta", "Tokyo", "wuhan"]
print(df)
```

```python
df["marks"]=df["marks"]+20
print(df)
df.drop("City", axis=1)
```
